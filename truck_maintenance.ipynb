```python
import numpy as np
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
df=pd.read_csv("truck_maintenance_dataset.csv")
df.isnull().sum()
df["Mileage_km"]=df["Mileage_km"].fillna(df["Mileage_km"].mean())
df["Load_Weight_Tons"]=df["Load_Weight_Tons"].fillna(df["Load_Weight_Tons"].mean())
df=df.drop_duplicates()
df["Engine_Hours"]=df["Engine_Hours"].fillna(df["Engine_Hours"].mean())
df["Fuel_Type"]=df["Fuel_Type"].fillna(df["Fuel_Type"].mode()[0])
df=df.dropna(subset=["Maintenance_Needed"])
df["Maintenance_Needed"]=df["Maintenance_Needed"].map({"Yes":1,"No":0})
df=pd.get_dummies(df,columns=["Fuel_Type"],drop_first=True,dtype=int)
df=df.drop("Truck_ID",axis=1)
x=df.drop("Maintenance_Needed",axis=1)
y=df["Maintenance_Needed"]
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
model=LogisticRegression(max_iter=1000)
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy:", accuracy)
print("Classification Report:\n", classification_report(y_test,y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred))



```
